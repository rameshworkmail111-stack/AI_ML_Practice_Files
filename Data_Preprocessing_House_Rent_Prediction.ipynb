```python
numerical_columns = df.select_dtypes(include="number").columns
print(numerical_columns)
```

```python
categorical_columns = df.select_dtypes(exclude="number").columns
print(categorical_columns)
```

```python
X = df.drop(columns=["Price", "Id"])
y = df["Price"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("X:")
print(X.head())
```

```python
print("y:")
print(y.head())
print("-----------------------------------")
print(y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.20, train_size = 0.80, random_state = 34)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
for col in categorical_columns:
    print(f"\n---------------------------- {col} ------------------------------")
    print(df[col].value_counts())
```

```python
# Encoding

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
) 
```

```python
X_train_categories = encoder.fit_transform(X_train[categorical_columns])
print(X_train_categories)
```

```python
X_test_categories = encoder.transform(X_test[categorical_columns]) 
print(X_test_categories)
```

```python
# Feature scaling 

numerical_columns = [
    "Area",
    "Bedrooms",
    "Bathrooms",
    "Floors",
    "YearBuilt"
] 

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_numerical = scaler.fit_transform(
    X_train[numerical_columns]
)

X_test_numerical = scaler.transform(
    X_test[numerical_columns]
)

print("X_train_numerical shape:", X_train_numerical.shape)
print("X_test_numerical shape :", X_test_numerical.shape)
```
