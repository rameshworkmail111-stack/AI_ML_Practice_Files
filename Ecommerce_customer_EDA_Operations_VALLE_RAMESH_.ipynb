```python
# PROBLEM STATEMENT 1     -        CUSTOMER LOYALTY POINTS VS REPEAT PURCHASES

print("PROBLEM STATEMENT 1 - CUSTOMER LOYALTY POINTS VS REPEAT PURCHASES")

# Step 1: Select Required Columns

loyalty_df = df[[ "Customer_ID", "Customer_Loyalty_Points", "Previous_Orders" ]].copy()

print("Selected Data:")
print(loyalty_df.head())
print("------------------------------------------------------------------------------------------------------------------")

# Step 2: Customer-Level Analysis

customer_loyalty = loyalty_df.groupby( "Customer_ID").agg(
    Loyalty_Points=( "Customer_Loyalty_Points", "mean" ),
    Previous_Orders=( "Previous_Orders", "max" )).reset_index()

print("Customer-Level Analysis:")
print(customer_loyalty.head())

print("--------------------------------------------------------------------------------------------")

# Step 3: Calculate Correlation

correlation = customer_loyalty[ "Loyalty_Points"].corr( customer_loyalty[ "Previous_Orders" ])

print("\nCorrelation between Loyalty Points and Previous Orders: ", round(correlation, 3))

print("-----------------------------------------------------------------------------------------------")

# Step 4: Interpret Correlation

print("\nCorrelation Interpretation:")

if correlation >= 0.7:

    print(
        "Strong Positive Relationship"
    )

elif correlation >= 0.3:

    print(
        "Moderate Positive Relationship"
    )

elif correlation > 0:

    print(
        "Weak Positive Relationship"
    )

elif correlation <= -0.7:

    print(
        "Strong Negative Relationship"
    )

elif correlation <= -0.3:

    print(
        "Moderate Negative Relationship"
    )

elif correlation < 0:

    print(
        "Weak Negative Relationship"
    )

else:

    print(
        "No Linear Relationship"
    )
```

```python
# Step 5: Create Loyalty Groups

customer_loyalty["Loyalty_Group"] = pd.qcut( 
    customer_loyalty[ "Loyalty_Points"],
    q=4,
    labels=[ "Low", "Medium", "High", "Very High"]
)

print("\nLoyalty Groups:")
print( customer_loyalty[[ "Customer_ID", "Loyalty_Points", "Previous_Orders", "Loyalty_Group"]].head(10))

print("---------------------------------------------------------------")

# Step 6: Analyze Previous Orders by Loyalty Group

loyalty_group_analysis = customer_loyalty.groupby(
    "Loyalty_Group",
    observed=True
)["Previous_Orders"].agg( [ "count", "mean", "median", "min","max" ]).sort_values(
    "mean",
    ascending=False
)


print("\nPrevious Orders by Loyalty Group:")
print(loyalty_group_analysis)
```

```python
# PROBLEM STATEMENT 2 - FACTORS AFFECTING CUSTOMER SATISFACTION

print("PROBLEM STATEMENT 2 - FACTORS AFFECTING CUSTOMER SATISFACTION")

# STEP 1: SELECT REQUIRED VARIABLES

satisfaction_df = df[
    [
        "Shipping_Days",
        "Discount_Pct",
        "Category",
        "Delivery_Rating",
        "Customer_Satisfaction"
    ]
].copy()

print("\nSelected Data:")
print(satisfaction_df.head())


print("----------------------------------------------------------------------------------------------------------------")
# STEP 2: CORRELATION ANALYSIS

correlation_matrix = satisfaction_df[
    [
        "Shipping_Days",
        "Discount_Pct",
        "Delivery_Rating",
        "Customer_Satisfaction"
    ]
].corr()


# Correlation of each numerical factor
# with Customer Satisfaction

satisfaction_correlation = (
    correlation_matrix["Customer_Satisfaction"]
    .drop("Customer_Satisfaction")
    .sort_values(
        key=abs,
        ascending=False
    )
)

print("\nCorrelation with Customer Satisfaction:")
print(satisfaction_correlation)
```

```python
# STEP 3: CATEGORY ANALYSIS

category_satisfaction = df.groupby(
    "Category"
)["Customer_Satisfaction"].agg(
    [
        "count",
        "mean",
        "median",
        "min",
        "max"
    ]
).sort_values(
    "mean",
    ascending=False
)

print("\nCustomer Satisfaction by Category:")
print(category_satisfaction)


print("----------------------------------------------------------------------------------------------------------------")
# STEP 4: IDENTIFY STRONGEST NUMERICAL FACTOR

strongest_factor = satisfaction_correlation.index[0]

strongest_correlation = satisfaction_correlation.iloc[0]

print("\nStrongest Numerical Factor:")
print(strongest_factor)

print(
    "Correlation:",
    round(strongest_correlation, 3)
)
```

```python
# PROBLEM STATEMENT 4 - DISCOUNT IMPACT ON BUSINESS PERFORMANCE

print("PROBLEM STATEMENT 4 - DISCOUNT IMPACT ON BUSINESS PERFORMANCE")

# STEP 1: SELECT REQUIRED VARIABLES

discount_df = df[
    [
        "Discount_Pct",
        "Net_Sales",
        "Profit",
        "Profit_Margin_Pct"
    ]
].copy()

print("\nSelected Data:")
print(discount_df.head())


# --------------------------------------------------------------------------
# STEP 2: CREATE DISCOUNT GROUPS

discount_df["Discount_Group"] = pd.cut(
    discount_df["Discount_Pct"],
    bins=[-1,5,10,15,20,25,30,35],
    labels=[
        "0-5%",
        "5-10%",
        "10-15%",
        "15-20%",
        "20-25%",
        "25-30%",
        "30-35%"
    ]
)
print("----------------------------------------------------------------------")
print("\nDiscount Groups:")
print(
    discount_df[
        [
            "Discount_Pct",
            "Discount_Group"
        ]
    ].head(10)
)
```

```python
# STEP 3: CALCULATE BUSINESS METRICS

discount_analysis = discount_df.groupby(
    "Discount_Group",
    observed=True
).agg(

    Average_Net_Sales=(
        "Net_Sales",
        "mean"
    ),

    Total_Net_Sales=(
        "Net_Sales",
        "sum"
    ),

    Average_Profit=(
        "Profit",
        "mean"
    ),

    Total_Profit=(
        "Profit",
        "sum"
    ),

    Average_Profit_Margin=(
        "Profit_Margin_Pct",
        "mean"
    )
)


print("\nDiscount Group Analysis:")
print(discount_analysis)

print("--------------------------------------------------------------------------")
# STEP 4: CORRELATION ANALYSIS

discount_correlation = discount_df[
    [
        "Discount_Pct",
        "Net_Sales",
        "Profit",
        "Profit_Margin_Pct"
    ]
].corr()

print("\nCorrelation Matrix:")
print(discount_correlation)


print("\nDiscount vs Net Sales:",
      round(
          discount_correlation.loc[
              "Discount_Pct",
              "Net_Sales"
          ],
          3
      )
)

print("Discount vs Profit:",
      round(
          discount_correlation.loc[
              "Discount_Pct",
              "Profit"
          ],
          3
      )
)

print("Discount vs Profit Margin:",
      round(
          discount_correlation.loc[
              "Discount_Pct",
              "Profit_Margin_Pct"
          ],
          3
      )
)
```

```python
# VISUALIZATIONS

fig, axes = plt.subplots( 1, 3,figsize=(18, 5))

# VISUALIZATION 1 - NET SALES

sns.barplot(
    data=discount_analysis.reset_index(),
    x="Discount_Group",
    y="Average_Net_Sales",
    ax=axes[0]
)

axes[0].set_title(
    "Average Net Sales by Discount"
)

axes[0].set_xlabel(
    "Discount Group"
)

axes[0].set_ylabel(
    "Average Net Sales"
)


# ------------------------------------------------
# VISUALIZATION 2 - PROFIT

sns.barplot(
    data=discount_analysis.reset_index(),
    x="Discount_Group",
    y="Average_Profit",
    ax=axes[1]
)

axes[1].set_title(
    "Average Profit by Discount"
)

axes[1].set_xlabel(
    "Discount Group"
)

axes[1].set_ylabel(
    "Average Profit"
)


# ------------------------------------------------
# VISUALIZATION 3 - PROFIT MARGIN

sns.barplot(
    data=discount_analysis.reset_index(),
    x="Discount_Group",
    y="Average_Profit_Margin",
    ax=axes[2]
)

axes[2].set_title(
    "Average Profit Margin by Discount"
)

axes[2].set_xlabel(
    "Discount Group"
)

axes[2].set_ylabel(
    "Average Profit Margin (%)"
)

plt.tight_layout()
plt.show()


# -------------------------------------------------------------------------------------
# FINAL RESULT

best_profit_group = discount_analysis[
    "Average_Profit"
].idxmax()

best_margin_group = discount_analysis[
    "Average_Profit_Margin"
].idxmax()

best_sales_group = discount_analysis[
    "Average_Net_Sales"
].idxmax()

print("FINAL RESULT:- \n")

print(
    "Highest Average Net Sales:",
    best_sales_group
)

print(
    "Highest Average Profit:",
    best_profit_group
)

print(
    "Highest Average Profit Margin:",
    best_margin_group
)
```

```python
# PROBLEM STATEMENT 5 - PRODUCT REVENUE & PROFITABILITY

print("PROBLEM STATEMENT 5 - PRODUCT REVENUE & PROFITABILITY")

# STEP 1: SELECT REQUIRED VARIABLES

product_df = df[
    [
        "Product_Name",
        "Net_Sales",
        "Profit",
        "Profit_Margin_Pct",
        "Quantity",
        "Order_ID"
    ]
].copy()

print("\nSelected Data:")
print(product_df.head())

# ----------------------------------------------------------------------------------
# STEP 2: GROUP BY PRODUCT

product_analysis = product_df.groupby(
    "Product_Name"
).agg(

    Total_Revenue=(
        "Net_Sales",
        "sum"
    ),

    Total_Profit=(
        "Profit",
        "sum"
    ),

    Average_Profit=(
        "Profit",
        "mean"
    ),

    Average_Profit_Margin=(
        "Profit_Margin_Pct",
        "mean"
    ),

    Total_Quantity=(
        "Quantity",
        "sum"
    ),

    Total_Orders=(
        "Order_ID",
        "nunique"
    )
)


print("\nProduct Analysis:")
print(product_analysis)

# ---------------------------------------------------------------------------
# STEP 3: RANK PRODUCTS

print("\nTop 10 Products by Revenue:")

top_revenue = product_analysis.sort_values(
    "Total_Revenue",
    ascending=False
).head(10)

print(top_revenue)
print("---------------------------------------------------------------------------")

print("\nTop 10 Products by Profit:")

top_profit = product_analysis.sort_values(
    "Total_Profit",
    ascending=False
).head(10)

print(top_profit)

print("---------------------------------------------------------------------------")
print("\nTop 10 Products by Profit Margin:")

top_margin = product_analysis.sort_values(
    "Average_Profit_Margin",
    ascending=False
).head(10)

print(top_margin)

```

```python
# STEP 4: SET THRESHOLDS

revenue_threshold = product_analysis[
    "Total_Revenue"
].median()

profit_threshold = product_analysis[
    "Total_Profit"
].median()

margin_threshold = product_analysis[
    "Average_Profit_Margin"
].median()


print("\nRevenue Threshold:",
      revenue_threshold)

print("Profit Threshold:",
      profit_threshold)

print("Profit Margin Threshold:",
      margin_threshold)


# --------------------------------------------------------------------------------------------
# STEP 5: PRODUCT CLASSIFICATION

# High Revenue + Low Profit

high_revenue_low_profit = product_analysis[
    (product_analysis["Total_Revenue"] > revenue_threshold)
    &
    (product_analysis["Total_Profit"] < profit_threshold)
]

print("\nHigh Revenue + Low Profit:")
print(
    high_revenue_low_profit.sort_values(
        "Total_Revenue",
        ascending=False
    )
)


# ------------------------------------------------
# High Revenue + High Profit

high_revenue_high_profit = product_analysis[
    (product_analysis["Total_Revenue"] > revenue_threshold)
    &
    (product_analysis["Total_Profit"] > profit_threshold)
]

print("\nHigh Revenue + High Profit:")
print(
    high_revenue_high_profit.sort_values(
        "Total_Profit",
        ascending=False
    )
)


# ------------------------------------------------
# High Profit Margin Products

high_margin_products = product_analysis[
    product_analysis["Average_Profit_Margin"]
    > margin_threshold
]

print("\nHigh Profit Margin Products:")
print(
    high_margin_products.sort_values(
        "Average_Profit_Margin",
        ascending=False
    )
)
```
