```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
import pandas as pd

df = pd.read_csv("student_dataset.csv")

df.head()
```

```python
# (1.) Lineplot 

sns.set_theme()                 # default theme of seaborn

plt.figure(figsize = (6,4.2)) 

sns.lineplot(data = df, x = "StudyHours", y = "Marks", color = "red", marker = "^") 

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Marks") 
plt.show()
```

```python
# (2.) Bar Graph

sns.set_theme()                 # default theme of seaborn
plt.subplots(figsize = (7,4.5)) # plt.figure(figsize = (7,4.5)) is same 

sns.barplot(data = df, x = "Branch", y = "Marks" , color = "#3179CB") 

plt.title("Students per branch")
plt.xlabel("Branch")
plt.ylabel("No.of students")
plt.show()
```

```python
# (3.) Histograms 

sns.set_theme()
plt.figure(figsize = (8,6))

sns.histplot(data = df, x = "Marks", bins = 6) 

plt.title("Distribution of marks")
plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.show()
```

```python
# (4.) Scatter plot 

sns.set_theme()

plt.figure(figsize = (10,7))
sns.scatterplot(data = df, x = "StudyHours", y = "Marks")

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Marks") 

plt.show()
```
