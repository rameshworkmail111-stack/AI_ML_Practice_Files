```python
# use write mode as a editor 
"""
Note:- 
when ever we need to edit an text inside string, we need to perform operation on same file, 
becasue when we open new file with write mode, it will rewrite everything, we can't edit on existing data.
"""
# i want to change the content form 14 th index.


with open("file.txt","w") as f:
    print(f.write("This class is           from calibo academy"))
    print(f.seek(14))
    print(f.write("python"))

with open("file.txt","r") as f:
    print(f.read())
```

```python
# better approach.
try:
    with open("file5.txt","x") as f:
        f.write("New file created using in create(x) method")

except FileExistsError:
    print("File already exists.")
```

```python
# read and write method (r+):- we can read an existing file, we can over write on existing file as well, but can not create new file.

# File must exist 
# Preserves content

with open("file5.txt","r+") as f:
    content = f.read()                          # read method 
    print(f"Original content: {content}") 

    new_content = f.write("New content written by r+ mode")            # write method
    print(new_content)

    f.seek(0)                   # help to read content from line. 
    New_content = f.read()    
    print ("New content: ", New_content)                    
    print()

```

```python
# write and read method (w+): perform both write and read operations, it creates new file, if not exists.

# Creates if missing
# Deletes old content

with open('file.txt', 'w+') as f:
    f.write('Hello World')      # Write method, replaces old content with nrew
    f.seek(0)                   # Move cursor to start
    print(f.read())             # Read: Hello World
```

```python
# Append and Read (a+ ): Creates file if missing, writes at END, reads from ANYWHERE 

with open('file.txt', 'a+') as f:
    f.write(' New data appending')           # Appends at end
    f.seek(0)                       # Move to start to read
    print(f.read())                 # Read everything
```
