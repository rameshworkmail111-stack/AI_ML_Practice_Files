```python
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)
print("Q1: ",Q1)
print("Q3: ",Q3)

IQR = Q3 - Q1 

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR 

print(lower_bound)
print(upper_bound)
print("----------------------------------")
print(df["Marks"].between(22.0,128.0).value_counts())
print("------------------------------------")
print(df["Marks"].mean())
print(df["Marks"].median())
print("------------------------------------")
df["Marks"] = df["Marks"].fillna(73.5)
df.loc[df["Marks"] > 100 , "Marks"] = 73.5
print(df["Marks"].between(17.875,132.875).value_counts())
```

```python
Q1 = df["Attendance"].quantile(0.25)
Q3 = df["Attendance"].quantile(0.75)
print("Q1: ",Q1)
print("Q3: ",Q3)

IQR = Q3 - Q1 

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR 

print(lower_bound)
print(upper_bound)
print("----------------------------------")
print(df["Attendance"].between(45.0,111.0).value_counts())
print("------------------------------------")
print(df["Attendance"].mean())
print(df["Attendance"].median())
print("------------------------------------")
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
print(df["Attendance"].between(45.0,111.0).value_counts())
```

```python
Q1 = df["Study_Hours"].quantile(0.25)
Q3 = df["Study_Hours"].quantile(0.75)
print("Q1: ",Q1)
print("Q3: ",Q3)

IQR = Q3 - Q1 

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR 

print(lower_bound)
print(upper_bound)
print("----------------------------------")
print(df["Study_Hours"].between(-5.5,14.5).value_counts())
print("------------------------------------")
print(df["Study_Hours"].mean())
print(df["Study_Hours"].median())
print("------------------------------------")
df["Study_Hours"] = df["Study_Hours"].fillna(5.0)
df.loc[df["Study_Hours"] > 14.5 , "Study_Hours"] = 5.0
print(df["Study_Hours"].between(-5.5,14.5).value_counts())
```

```python
# Graphs - Matplotlib 

# (1.) Line graph:- 
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize = (15,4))

x_corr = df["Student_ID"]
y_corr = df["Attendance"]
ax.plot(x_corr,y_corr, marker = "s", label = "Student_ID", color = "orange") 

ax.set_title("Student ID vs Attendance")
ax.set_xlabel("Student ID")
ax.set_ylabel("Attendance")
ax.legend()
plt.show()
```

```python
# (2.) Bar chart 

fig, ax = plt.subplots(figsize = (8,4))

Age_data = df["Age"].value_counts()

unique_age = Age_data.index 
age_counts = Age_data.values 

bars = ax.bar(unique_age, age_counts, color = "green", edgecolor = "blue" ) 

# Display count on top of each bar
ax.bar_label(bars)
ax.set_title("Age vs Count ")
ax.set_xlabel("Age")
ax.set_ylabel("Students Count")
plt.show()
```

```python
# (3.) Histogram 

fig, ax = plt.subplots(figsize = (7,4.5))

category = df["Marks"]

n, bins, patches = ax.hist(category, bins = 5, color = "red", edgecolor = "black")

# Display count on top of each bar 
ax.bar_label(
    patches,
    labels=[str(int(value)) for value in n],
    padding=3
)

ax.set_title("Marks Distribution")
ax.set_xlabel("Marks")
ax.set_ylabel("Count of students in the range")

plt.legend()
plt.show()
```

```python
# (4.) scatter plot

fig, ax = plt.subplots(figsize = (12,4.5))

x_corr = df["Attendance"]
y_corr = df["Marks"]

ax.scatter(x_corr,y_corr, color = "orange", alpha = 0.8)

ax.set_title("Marks")
ax.set_xlabel("Attendance")
ax.set_ylabel("Correlation between Attendance and Marks")
```

```python
# (5.) Pie chart 

fig, ax = plt.subplots(figsize = (8,8))

Department_data = df["Department"].value_counts()

Department = Department_data.index
Department_count = Department_data.values

ax.pie(Department_count, labels = Department, autopct = "%1.1f%%")

ax.set_title("Total Departments from college")
ax.set_xlabel("Department")
ax.set_ylabel("Students count")
```

```python
# (6.) subplots 

fig, axes = plt.subplots(1,2, figsize = (14,7)) 

axes[0].scatter(df["Study_Hours"],df["Marks"], color = "red", alpha = 0.7 )

axes[0].set_title("Correlation between Study Hours vs Marks")
axes[0].set_xlabel("Study Hours from students")
axes[0].set_ylabel("Marks")


n, bins, patches = axes[1].hist(df["Internet_Usage_Hours"],bins = 6, color = "blue", edgecolor = "white")

axes[1].bar_label(
    patches,
    labels=[str(int(value)) for value in n],
    padding=3
)

axes[1].set_title("Internet usage by students")
axes[1].set_xlabel("Internet packages")
axes[1].set_ylabel("Number of students usage")

plt.savefig("subplots.png",dpi = 300)

plt.legend()
plt.show()
```
