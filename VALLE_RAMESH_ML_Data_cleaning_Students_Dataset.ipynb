```python
# 22. dropna() = Removes rows or columns containing missing (NaN) values.

df = pd.read_csv("student_marks.csv")

print(df.dropna()) #Rows containing NaN values are removed.
print(df.dropna(axis=1)) #Remove Columns with NaN

"""Corner Cases
Default removes rows.
Use how="all" to remove rows only if all values are missing.
Use thresh= to specify the minimum number of non-null values required."""
```

```python
# (31.) loc[] = loc[] is used to select rows and columns by their labels/names.        loc → label-based selection.

print(df.loc[2])             # Selects the row whose index label is 2.
print("--------------------------------")
print(df.loc[[1,3,0]])         # Selects the rows whose index label is 1,3,0.
print("--------------------------------")
print(df.loc[1:5, ["StudentID", "Marks"]])                    #Rows    → 1 to 5,            Columns → Name, Marks 
print("---------------------------------")

# Updating values => loc is also very useful for modifying specific data:
 
df.loc[df["Marks"] > 90, "Result"] = "A+"
print(df[["Marks","Result"]])

# corner cases
# (1.) df.loc[20],       selects the row with index label 20, It does NOT mean "20th row".
# (2.) df.loc[1:5],      selects 1, 2, 3, 4, 5.   This differs from normal Python slicing.
# df.loc[100]             If index 100 doesn't exist, you'll get a KeyError.
```

```python
# (32.) iloc[] = iloc[] selects rows and columns using their integer positions.                     iloc → integer-location based selection

print(df.iloc[0])                # selects the first row, regardless of its index label.
print("---------------------------------")
print(df.iloc[[0, 2]])             # First row, Third row
print("---------------------------------")

# Select rows and columns

print(df.iloc[0:3, 0:2])                 # Rows    → positions 0, 1, 2   &  Columns → positions 0, 1       ⚠️ Here Python-style slicing applies, so the ending position is excluded.
print("---------------------------------")
print(df.iloc[2, 1])                      # 3rd row, 2nd column
```

```python
# (1.) Line graph:- 
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize = (18,4.5))

x_corr = df["StudentID"]
y_corr = df["StudyHours"]
ax.plot(x_corr,y_corr, marker = "^", label = "Student_ID", color = "gray") 

ax.set_title("Student ID vs StudyHours")
ax.set_xlabel("Student ID")
ax.set_ylabel("Study Hours")
ax.legend()
plt.show()
```

```python
# (2.) Bar chart 

fig, ax = plt.subplots(figsize = (8,4))

Branch_data = df["Branch"].value_counts()

unique_branch = Branch_data.index 
branch_counts = Branch_data.values 

bars = ax.bar(unique_branch,branch_counts, color = "red", edgecolor = "black" ) 

# Display count on top of each bar
ax.bar_label(bars)
ax.set_title("Branch vs Count ")
ax.set_xlabel("Branch")
ax.set_ylabel("Students Count")
plt.show()
```

```python
# (3.) Histogram 

fig, ax = plt.subplots(figsize = (9,6))

category = df["Marks"]

n, bins, patches = ax.hist(category, bins = 7, color = "skyblue", edgecolor = "orange")

# Display count on top of each bar 
ax.bar_label(
    patches,
    labels=[str(int(value)) for value in n],
    padding=3
)

ax.set_title("Marks Distribution")
ax.set_xlabel("Marks")
ax.set_ylabel("Count of students in the range")

plt.legend()
plt.show()
```

```python
# (4.) scatter plot

fig, ax = plt.subplots(figsize = (9,4.5))

x_corr = df["StudyHours"]
y_corr = df["Marks"]

ax.scatter(x_corr,y_corr, color = "green", alpha = 0.7)

ax.set_title("Marks")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Correlation between Study Hours and Marks")
```

```python
# (5.) Pie chart 

fig, ax = plt.subplots(figsize = (9,7))

Branch_data = df["Branch"].value_counts()

Branch = Branch_data.index
Branch_count = Branch_data.values

ax.pie(Branch_count, labels = Branch, autopct = "%1.1f%%")

ax.set_title("Total Branches from college")
ax.set_xlabel("Branches")
ax.set_ylabel("Students count")
```

```python
# (6.) subplots 

fig, axes = plt.subplots(1,2, figsize = (15,6)) 

axes[0].scatter(df["AttendancePercent"],df["Marks"], color = "blue", alpha = 0.8 )

axes[0].set_title("Correlation between Attendance vs Marks")
axes[0].set_xlabel("Attendance from students")
axes[0].set_ylabel("Marks")


n, bins, patches = axes[1].hist(df["AttendancePercent"],bins = 7, color = "orange", edgecolor = "skyblue")

axes[1].bar_label(
    patches,
    labels=[str(int(value)) for value in n],
    padding=3
)

axes[1].set_title("Attendance by students")
axes[1].set_xlabel("Attendance ranges")
axes[1].set_ylabel("Number of students usage")

plt.savefig("subplots.png",dpi = 300)

plt.legend()
plt.show()
```
