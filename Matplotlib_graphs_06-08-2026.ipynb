```python
from matplotlib import pyplot as plt   
```

```python
import seaborn as sns
```

```python
import pandas as pd
```

```python
df = pd.read_csv("student_dataset.csv")

print(df.head())
```

```python
# (1.) Line Graph 

days = list(range(1,11))
avg = [62, 64, 65, 68, 70, 71, 74, 75, 77, 78]

fig, ax = plt.subplots(
    figsize = (7,4.2) )

ax.plot(days, avg, marker = "o", color = "#065A82")

ax.set_title(
    "Avg Marks Over 10 Tests") 
ax.set_xlabel("Test Number")
ax.set_ylabel("Average Marks")
```

```python
# (2.) Bar chart

counts = df ["Branch"].value_counts()

fig, ax = plt.subplots(
    figsize = (7,4.2) )

ax.bar(counts.index, counts.values, color = "#1c7293")

ax.set_title(
    "Student per branch") 
ax.set_xlabel("Branch")
ax.set_ylabel("No.of students")
```

```python
# (3.) Scatter plot 

fig, ax = plt.subplots(figsize = (7,4.2))

ax.hist(df["Marks"],bins=5, color="green", edgecolor= "black" )

ax.set_title("Distribution of marks")
ax.set_xlabel("Marks")
ax.set_ylabel("Number of students")
plt.show()
```

```python
# (4.) scatter plot 

fig, ax = plt.subplots(figsize = (7,4.2)) 
ax.scatter(df["StudyHours"],df["Marks"], color = "red", marker = "^", alpha = 0.8 )

ax.set_title("Study Hours vs Marks")
ax.set_xlabel("Study hours per hour")
ax.set_ylabel("Marks")
plt.show()
```

```python
# (5.) Pie chart

fig, ax = plt.subplots(figsize = (7,4.2))
counts = df["Branch"].value_counts() 
categories = counts.index 
percentages = counts.values 

ax.pie(percentages, labels=categories, autopct="%1.1f%%")
ax.set_title("Branch Distribution")
plt.show()
```

```python
# (6.) Sub plots

fig, axes = plt.subplots(1, 2, figsize = (9,4.2) )

axes[0].hist(df["Marks"], bins = 6, color = "violet", edgecolor = "black")
axes[0].set_title("Marks Distribution")

axes[1].scatter(df["StudyHours"],df["Marks"], color = "blue", alpha = 0.7 )
axes[1].set_title("StudyHours vs Marks")

plt.tight_layout()
plt.show()
```

```python
# (1.) Legend

days = list(range(1,11))
avg = [62, 64, 65, 68, 70, 71, 74, 75, 77, 78]

days_2 = list(range(11,21))
avg_2 = [62, 64, 65, 68, 70, 71, 74, 75, 77, 78]

fig, ax = plt.subplots(
    figsize = (7,4.2) )

ax.plot(days, avg, marker = "o", color = "#065A82", label = "1st semester", linestyle = "dashed")

ax.plot(days_2, avg_2, marker = "*", color = "#2BE125", label = "2nd semester", linewidth = 2)

ax.grid(
    linestyle="--",
    alpha=0.5
)

ax.set_title(
    "Avg Marks Over 10 Tests") 
ax.set_xlabel("Test Number")
ax.set_ylabel("Average Marks")
ax.legend()
plt.savefig("customization_graph.png",dpi=300, bbox_inches = "tight")
```
