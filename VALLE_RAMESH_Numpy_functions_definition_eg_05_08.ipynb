```python
import numpy as np
```

```python
A = np.array([[3, 2, 1, 4],
              [5, 2, 1, 6]])

B = np.array([[3, 2, 1, 4],
              [5, 2, 0, 6]])

C = np.array([[True, False, False],
              [True, True, True]])

D = np.array([0.1, 0.3])
```

```python
# (1.) np.all() = Return True if all elements of the array evaluate to True, otherwise return False.

print(np.all(A))
print(np.all(B))
```

```python
# cornercases

# (1.) Empty array 
print(np.all(np.array([])))

print("-----------------------------------------------------")

# (2.) Array with NaN values
print(np.all(np.array([1, 2, np.nan])))

print("-----------------------------------------------------")

# Practice

for name, array in zip(list('ABCD'), [A, B, C, D]):
    print(f'{name}: {np.all(array)}')

print("-----------------------------------------------------")

for name, array in zip(list('ABC'), [A, B, C]):
    print(f'{name}: {np.all(array, axis=1)}')

print("-----------------------------------------------------")

for name, array in zip(list('ABC'), [A, B, C]):
    print(f'{name}: {np.all(array, axis=0)}')
```

```python
# (2.) np.any() = Return True if any of the elements of the array evaluate to True, otherwise return False.

A = np.array([[3, 2, 1, 4],
              [5, 2, 1, 6]])

B = np.array([[3, 2, 1, 4],
              [5, 2, 0, 6]])

C = np.array([[True, False, False],
              [True, True, True]])

D = np.array([0.1, 0.3])

for name, array in zip(list('ABCD'), [A, B, C, D]):
    print(f'{name}: {np.any(array)}')
```

```python
# (63.) np.polyval() = Evaluate a polynomial at specific values.

coefficients = [1, -6, 11, -6]  # Represents the polynomial x^3 - 6x^2 + 11x - 6
x_values = [0, 1, 2, 3]
y_values = np.polyval(coefficients, x_values)
print(y_values)  # Output: [-6. 0. 0. 0.]
```

```python
# (64.) np.polyadd() = Add two polynomials represented by their coefficients.

coefficients1 = [1, -6, 11, -6]  # Represents the polynomial x^3 - 6x^2 + 11x - 6
coefficients2 = [1, 0, 0]        # Represents the polynomial x
sum_coefficients = np.polyadd(coefficients1, coefficients2)
print(sum_coefficients)  # Output: [2. -6. 11. -6.]
```

```python
# (65.and) np.polysub() = Subtract two polynomials represented by their coefficients.

coefficients1 = [1, -6, 11, -6]  # Represents the polynomial x^3 - 6x^2 + 11x - 6
coefficients2 = [1, 0, 0]        # Represents the polynomial x
diff_coefficients = np.polysub(coefficients1, coefficients2)
print(diff_coefficients)  # Output: [0. -6. 11. -6.]
```

```python
# (66.) np.polymul() = Multiply two polynomials represented by their coefficients.

coefficients1 = [1, -6, 11, -6]  # Represents the polynomial x^3 - 6x^2 + 11x - 6
coefficients2 = [1, 0, 0]        # Represents the polynomial x
product_coefficients = np.polymul(coefficients1, coefficients2)
print(product_coefficients)  # Output: [1. -6. 11. -6.]
```

```python
# np.polydiv() = Divide two polynomials represented by their coefficients.

coefficients1 = [1, -6, 11, -6]  # Represents the polynomial x^3 - 6x^2 + 11x - 6
coefficients2 = [1, 0, 0]        # Represents the polynomial x
quotient_coefficients, remainder_coefficients = np.polydiv(coefficients1, coefficients2)
print(quotient_coefficients)  # Output: [1. -6. 11.]
print(remainder_coefficients)  # Output: [-6.]
```

```python
# (70.) np.char.split() = Split each string in the array by a specified delimiter.
A = np.array(['Hello,World', 'Python,Programming'])
print(np.char.split(A, ','))  # Output: [['Hello', 'World'] ['Python', 'Programming']]
```

```python
# (71.) np.char.strip() = Return a new array of strings with leading and trailing whitespace removed from each string.

A = np.array(['  Hello  ', '  World  '])
print(np.char.strip(A))  # Output: ['Hello' 'World']
```

```python
# (72.) np.char.replace() = Return a new array of strings with all occurrences of a substring replaced by another substring.

A = np.array(['Hello World', 'Python Programming'])
print(np.char.replace(A, ' ', '_'))  # Output: ['Hello_World' 'Python_Programming']
```

```python
# (73.) np.char.count() = Return a new array of integers with the number of occurrences of a substring in each string.

A = np.array(['Hello World', 'Python Programming'])
print(np.char.count(A, 'o'))  # Output: [2 1]
```

```python
# (74.) np.char.startswith() = check if string specified prefix or not 

A = np.array(['Hello World', 'Python Programming'])
print(np.char.startswith(A,"P"))
```
